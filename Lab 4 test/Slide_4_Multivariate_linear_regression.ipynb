```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# =========================================
# Load Dataset
# =========================================

df = pd.read_csv("multi_output_student_Multivariate_Linear_Regression_dataset.csv")

print("===== Dataset =====")
print(df)

# =========================================
# Define Input and Output
# =========================================

print("\n===== Defining X and Y =====")

# Multiple input variables
X = df[["Hours", "Sleep"]]

# Multiple output variables
y = df[["Math", "Science"]]

print("\nX Values:")
print(X)

print("\ny Values:")
print(y)

# =========================================
# Create Model
# =========================================

print("\n===== Creating Model =====")

model = LinearRegression()

# =========================================
# Train Model
# =========================================

print("\n===== Training Model =====")

model.fit(X, y)

# =========================================
# Predict Outputs
# =========================================

print("\n===== Predicting Math and Science Marks =====")

prediction = model.predict([[7, 8]])

print("Predicted Marks:")
print(prediction)

# =========================================
# Plotting Section
# =========================================

print("\n===== Showing Scatter Plots =====")

# Scatter plot for Hours vs Math
plt.scatter(df["Hours"], df["Math"])

plt.xlabel("Study Hours")
plt.ylabel("Math Marks")

plt.title("Hours vs Math Marks")

plt.show()

# =========================================

# Scatter plot for Hours vs Science
plt.scatter(df["Hours"], df["Science"])

plt.xlabel("Study Hours")
plt.ylabel("Science Marks")

plt.title("Hours vs Science Marks")

plt.show()
```
