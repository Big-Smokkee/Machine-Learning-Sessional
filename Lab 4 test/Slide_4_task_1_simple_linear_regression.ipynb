```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# =========================================
# Load Dataset
# =========================================

df = pd.read_csv("student_marks.csv")

print("===== Dataset =====")
print(df)

# =========================================
# Define Input and Output
# =========================================

print("\n===== Defining X and Y =====")

X = df[["Hours"]]   # Independent variable
y = df["Marks"]     # Dependent variable

print("X Values:")
print(X)

print("\ny Values:")
print(y)

# =========================================
# Create Model
# =========================================

print("\n===== Creating Linear Regression Model =====")

model = LinearRegression()

# =========================================
# Train Model
# =========================================

print("\n===== Training Model =====")

model.fit(X, y)

# =========================================
# Predict Value
# =========================================

print("\n===== Predicting Marks =====")

predicted_marks = model.predict([[7]])

print("Predicted Marks for 7 hours study:")
print(predicted_marks)

# =========================================
# Plotting Section
# =========================================

print("\n===== Showing Scatter Plot and Regression Line =====")

# Scatter plot (actual data points)
plt.scatter(df["Hours"], df["Marks"])

# Regression line
plt.plot(df["Hours"], model.predict(X))

# Labels
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Title
plt.title("Simple Linear Regression")

# Show graph
plt.show()
```
