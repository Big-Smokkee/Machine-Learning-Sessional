```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# =========================================
# Load Dataset
# =========================================

df = pd.read_csv("student_data_for_multiple_linear_regression.csv")

print("===== Dataset =====")
print(df)

# =========================================
# Define Input and Output
# =========================================

print("\n===== Defining X and Y =====")

# Multiple input columns
X = df[["Hours", "Sleep", "Attendance"]]

# Output column
y = df["Marks"]

print("\nX Values:")
print(X)

print("\ny Values:")
print(y)

# =========================================
# Create Model
# =========================================

print("\n===== Creating Model =====")

model = LinearRegression()

# =========================================
# Train Model
# =========================================

print("\n===== Training Model =====")

model.fit(X, y)

# =========================================
# Predict Marks
# =========================================

print("\n===== Predicting Marks =====")

prediction = model.predict([[7, 8, 95]])

print("Predicted Marks:")
print(prediction)
```
