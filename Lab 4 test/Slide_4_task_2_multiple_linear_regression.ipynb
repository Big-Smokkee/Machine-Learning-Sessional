```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# =========================================
# Load Dataset
# =========================================

df = pd.read_csv("student_data_for_multiple_linear_regression.csv")

print("===== Dataset =====")
print(df)

# =========================================
# Define Input and Output
# =========================================

print("\n===== Defining X and Y =====")

# Multiple input columns
X = df[["Hours", "Sleep", "Attendance"]]

# Output column
y = df["Marks"]

print("\nX Values:")
print(X)

print("\ny Values:")
print(y)

# =========================================
# Create Model
# =========================================

print("\n===== Creating Model =====")

model = LinearRegression()

# =========================================
# Train Model
# =========================================

print("\n===== Training Model =====")

model.fit(X, y)

# =========================================
# Predict Marks
# =========================================

print("\n===== Predicting Marks =====")

prediction = model.predict([[7, 8, 95]])

print("Predicted Marks:")
print(prediction)

# =========================================
# Plotting Section
# =========================================

print("\n===== Showing Scatter Plot =====")

# Scatter Plot using Hours vs Marks
plt.scatter(df["Hours"], df["Marks"])

# Regression line
plt.plot(df["Hours"], model.predict(X))

# Labels
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Title
plt.title("Hours vs Marks")

# Show graph
plt.show()
```
