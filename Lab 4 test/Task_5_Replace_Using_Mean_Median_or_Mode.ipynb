```python
import pandas as pd

# Load dataset
df = pd.read_csv("employee - Sheet1.csv")

# Show original dataset
print("===== Original Data =====")
print(df)

# =========================================
# Replace Missing Salary using Mean
# =========================================

print("\n===== Replacing Missing Salary Using Mean =====")

# Calculate mean salary
mean_salary = df["Salary"].mean()

print("Mean Salary is:", mean_salary)

# Replace missing salary values
df["Salary"] = df["Salary"].fillna(mean_salary)

# Show updated dataframe
print(df)

# =========================================
# Replace Missing Age using Median
# =========================================

print("\n===== Replacing Missing Age Using Median =====")

# Calculate median age
median_age = df["Age"].median()

print("Median Age is:", median_age)

# Replace missing age values
df["Age"] = df["Age"].fillna(median_age)

# Show updated dataframe
print(df)

# =========================================
# Replace Missing City using Mode
# =========================================

print("\n===== Replacing Missing City Using Mode =====")

# Calculate mode city
mode_city = df["City"].mode()[0]

print("Mode City is:", mode_city)

# Replace missing city values
df["City"] = df["City"].fillna(mode_city)

# Show final dataframe
print(df)

print("\n===== Final Cleaned Data =====")
print(df)
```
