```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = pd.read_csv("multivariateLinearRegression.csv")
print("Dataset")
print(data)

X = data[["hours","sleep"]]
Y = data[['math', 'science']]

print("X values: ")
print(X)
print("Y values: ")
print(Y)

model = LinearRegression()
model.fit(X, Y)
predictionResult = model.predict([[6,8]])
print("Prediction result: ", predictionResult)

plt.xlabel("Study Hours")
plt.ylabel("Math Marks")
plt.title("Math Marks VS Study hours")
plt.scatter(data["hours"], data["math"])
plt.show()

plt.xlabel("Study Hours")
plt.ylabel("Science Marks")
plt.title("Science Marks VS Study hours")
plt.scatter(data["hours"], data["science"])
plt.show()
```
