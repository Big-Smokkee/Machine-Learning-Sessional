```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


data = pd.read_csv('simpleLinearRegression.csv')
print("====Dataset====")
print(data)

print("Defining X and Y")
X = data[["studyHours"]]
Y = data["marks"]

print("X values")
print(X)
print("Y values")
print(Y)

#Create Model
print("===Creating Linear Regression Model===")
model = LinearRegression()
print("===Training Linear Regression Model===")
model.fit(X, Y)

# Predicting Output of study hour 9
print("===Predicting Output of study hour 9===")
predictedMarks = model.predict([[9]])
print("Predicted Marks for study hour 9: ", predictedMarks)

plt.title("Simple Linear Regression Graph")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Scatter plot (actual data points)
plt.scatter(data["studyHours"], data["marks"])

# Regression line
plt.plot(data["studyHours"], model.predict(X))
plt.scatter(X, Y)
plt.show()

```
