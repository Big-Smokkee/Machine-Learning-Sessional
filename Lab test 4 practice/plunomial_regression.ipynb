```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# =========================================
# Load Dataset
# =========================================

df = pd.read_csv("student_poly.csv")

print("===== Dataset =====")
print(df)

# =========================================
# Define X and Y
# =========================================

X = df[["studyHours"]]
y = df["marks"]

# =========================================
# Convert X into Polynomial Features
# =========================================

print("\n===== Creating Polynomial Features =====")

poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

print("Transformed X:")
print(X_poly)

# =========================================
# Create Model
# =========================================

model = LinearRegression()

# =========================================
# Train Model
# =========================================

model.fit(X_poly, y)

# =========================================
# Prediction
# =========================================

predicted = model.predict(poly.transform([[7]]))

print("\nPredicted Marks for 7 hours study:")
print(predicted)

# =========================================
# Visualization
# =========================================

print("\n===== Plotting Polynomial Curve =====")

# Sort X for smooth curve
X_sorted = X.sort_values(by="studyHours")

# Predict for curve line
y_curve = model.predict(poly.transform(X_sorted))

# Scatter plot (actual data)
plt.scatter(X, y)

# Curve line
plt.plot(X_sorted, y_curve)

plt.xlabel("studyHours")
plt.ylabel("Marks")

plt.title("Polynomial Regression (Degree 2)")

plt.show()
```
